## Linear fit of PWM vs Thrust in grams measured empirically on the handmade thrust stand


```python
import numpy as np

# Pwm values 
pwm = np.array([
    0, 1000, 2000, 3000, 4000, 5000, 6000, 7000, 8000, 9000, 10000, 11000,
    12000, 13000, 14000, 15000, 16000, 17000, 18000, 19000, 20000, 21000,
    22000, 23000, 24000, 25000, 26000, 27000, 28000, 29000, 30000, 31000,
    32000, 33000, 34000, 35000, 36000, 37000, 38000, 39000, 40000, 41000,
    42000, 43000, 44000, 45000, 46000, 47000, 48000, 49000, 50000, 51000,
    52000, 53000, 54000, 55000, 56000, 57000, 58000, 59000, 60000
])

# Measured thrust
grams = np.array([
    0.0, 0.22, 0.3, 0.68, 1.24, 1.9, 2.68, 3.31, 4.24, 5.01, 5.88, 6.65,
    7.48, 8.26, 9.12, 9.97, 10.82, 11.53, 12.59, 13.27, 14.18, 15.08,
    15.64, 16.78, 17.49, 18.36, 19.03, 19.89, 20.9, 21.62, 22.32, 23.23,
    24.01, 24.81, 25.67, 26.27, 26.85, 27.94, 28.75, 29.56, 30.8, 31.39,
    32.4, 33.5, 34.27, 35.1, 35.9, 36.7, 37.5, 38.3, 39.0, 40.0,
    40.8, 42.0, 43.0, 44.0, 45.06, 45.8, 46.7, 47.0, 47.0
])

coeffs_lin = np.polyfit(pwm, grams, 1)
m, k = coeffs_lin
grams_lin = np.polyval(np.array([7.915e-04, -2]), pwm)
r2_lin = 1 - np.sum((grams - grams_lin)**2) / np.sum((grams - np.mean(grams))**2)
print(f"\nLinear fit R² = {r2_lin:.6f}  (y = {m:.6e}*pwm + {k:.4f})")
```

The function using the parameters to go from thrust in grams to pwm to convert mpc output to correct firmware motor input

```python
def thrust_to_pwm_linear(thrust: float) -> int: 
        # Convert to grams
        thrust_grams = thrust / 9.81 * 1000

        # Compute pwm using the linear fit coefficients
        pwm = (thrust_grams + 2.1948) / 8.271227e-04

        # Clip to cmdVel() logical values
        return int(np.clip(pwm, 0, 60000))

thrust_to_pwm_linear(0.03899 * 9.81)
```
